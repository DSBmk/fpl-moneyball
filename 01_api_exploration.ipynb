### Objective
Explore the Fantasy Premier League API and understand how its data can be accessed with Python.

### Analytical question
What player information is available through the FPL API that could eventually help us make better FPL decisions?

### First requests

```python
# Importing libraries
import requests
import pandas as pd
```

```python
# REQUEST: Building the request
url = "https://fantasy.premierleague.com/api/bootstrap-static/"
response = requests.get(url)
```

```python
# CHECK: Investigating what happened
display(type(response))
display(response.status_code)
if response.status_code == 200: 
    print('HTTP status code: success.') 
else: 
    print('Something went wrong!')
display(response.ok)
```

### Meet JSON

```python
# Converting the response into JSON
data = response.json()
```

```python
# CHECK: Investigating what happened
display(type(data))
display(data.keys())
```

```python
# Studying keys
print(f"Data type: {type(data['elements'])}")
print(f"Length: {len(data['elements'])}")
display(data["elements"][0])
```

`elements` contain players data.

```python
# Studying keys
print(f"Data type: {type(data['teams'])}")
print(f"Length: {len(data['teams'])}")
display(data["teams"][0])
```

`teams` contain teams data.

```python
# Studying keys
print(f"Data type: {type(data['events'])}")
print(f"Length: {len(data['events'])}")
display(data["events"][0])
```

`events` contain gameweek data.

```python
# Studying keys
print(f"Data type: {type(data['element_types'])}")
print(f"Length: {len(data['element_types'])}")
display(data["element_types"][0])
```

`element_types` contain player position data.

### Create the first DataFrame

```python
# Create the first DataFrame
df_players = pd.DataFrame(data["elements"])
```

```python
# Show all columns
pd.set_option('display.max_columns', None)
# Show all rows
pd.set_option('display.max_rows', None)
```

```python
# Explore the dataframe containing players data
display(df_players.head())
display(df_players.columns)
display(df_players.info())
```

## 03 - Build the Analytical Table
[Back to Top](#top)

### Select useful variables

```python
# Select useful columns
columns = [
    "id", "web_name", "team", "element_type",
    "now_cost", "total_points", "minutes",
    "form", "selected_by_percent"
]
```

```python
# Create a new DataFrame without changing the original
df_analysis = df_players[columns].copy()
df_analysis.head()
```

Raw data and analytical data should remain separate. We preserve the original dataset so we can revisit our decisions.

### Rename columns

```python
# Rename columns to make them easier to read
df_analysis = df_analysis.rename(columns={
    "web_name": "player_name",
    "element_type": "position_id",
    "team": "team_id"
})
```

```python
# Check the result
df_analysis.columns.tolist()
```

### Map IDs to readable names

```python
# Create a lookup dictionary for teams
team_lookup = df_teams.set_index("id")["name"].to_dict()
```

```python
# Check the result
list(team_lookup.items())[:5]
```

```python
# Add actual team names to the analytical table
df_analysis["team_name"] = df_analysis["team_id"].map(team_lookup)
```

```python
# Create a lookup dictionary for positions
position_lookup = df_positions.set_index("id")["singular_name"].to_dict()
```

```python
# Check the result
list(position_lookup.items())
```

```python
# Add actual position names to the analytical table
df_analysis["position_name"] = df_analysis["position_id"].map(position_lookup)
```

## Validate the analytical table

```python
# Check the analytical table
display(df_analysis.shape)
display(df_analysis.head())
```

```python
# Look for missing values
df_analysis[["team_name", "position_name"]].isna().sum()
```

```python
# Look for duplicates
df_analysis["id"].duplicated().sum()
```

Always validate transformations. A successful Python operation does not guarantee a correct analytical result.
